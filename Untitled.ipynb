```python
import os
import qrcode
from PIL import Image, ImageDraw


# Auto-Increment ID के लिए Counter
student_counter = 101


def generate_auto_id_card(student_name, student_class):
    global student_counter

    # 1. Automatic ID/Token Generation
    student_id = f"AMS-{student_counter}"
    student_counter += 1  # Agle bache ke liye ID apne aap 102, 103 ho jayegi

    width, height = 450, 270
    card = Image.new("RGB", (width, height), color=(255, 255, 255))
    draw = ImageDraw.Draw(card)

    # 2. Stylish Header
    draw.rectangle([(0, 0), (width, 55)], fill="black")
    draw.text((20, 12), "AYUSH MATHEMATICS CENTER", fill="white")
    draw.text((20, 32), "STUDENT IDENTIFICATION CARD", fill="white")

    # Design Lines
    draw.line([(0, 58), (width, 58)], fill="black", width=2)
    draw.rectangle([(15, 75), (20, 215)], fill="black")
    draw.line([(0, 230), (width, 230)], fill="black", width=1)

    # 3. Student Details (Auto Printed)
    draw.text((35, 75), f"ID NO      :  {student_id}", fill="black")
    draw.text((35, 110), f"NAME      :  {student_name}", fill="black")
    draw.text((35, 145), f"CLASS     :  {student_class}", fill="black")
    draw.text((35, 180), f"FEE/TEST  :  Rs 2.50", fill="black")

    # 4. Automatic QR Code Generation
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_M,
        box_size=6,
        border=1,
    )
    qr.add_data(student_id)
    qr.make(fit=True)
    qr_img = qr.make_image(fill_color="black", back_color="white")

    qr_img = qr_img.resize((110, 110))
    draw.rectangle([(310, 75), (428, 193)], outline="black", width=2)
    card.paste(qr_img, (314, 79))
    draw.text((328, 198), "SCAN FOR TEST", fill="black")

    draw.rectangle([(0, 0), (width - 1, height - 1)], outline="black", width=3)

    # Save Card
    if not os.path.exists("Student_ID_Cards"):
        os.makedirs("Student_ID_Cards")

    filename = f"Student_ID_Cards/{student_id}_{student_name}.png"
    card.save(filename)
    print(
        f"✅ Card Generated! ID: {student_id} | Name: {student_name} | Saved as: {filename}"
    )


# ==========================================
# आपको बस ऐसे नाम और क्लास देना है:
# ==========================================
generate_auto_id_card("Aayush Kumar", "Class 4")
generate_auto_id_card("Monika kumari", "Class 4")
generate_auto_id_card("Nishu kumari ", "Class 4")
```

```python
import os
from PIL import Image, ImageDraw
import qrcode
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas

# Auto-Increment ID Counter
student_counter = 101


# 1. ID Card Image Generator Function
def generate_single_card_image(student_name, student_class):
    global student_counter

    # Trim extra spaces from names
    student_name = student_name.strip()
    student_class = student_class.strip()

    student_id = f"AMTC-{student_counter}"
    student_counter += 1

    width, height = 500, 300
    card = Image.new("RGB", (width, height), color=(255, 255, 255))
    draw = ImageDraw.Draw(card)

    # Watermark AMTC
    draw.text((30, 80), "AMTC  AMTC  AMTC", fill=(235, 235, 235))
    draw.text((80, 140), "AMTC  AMTC  AMTC", fill=(235, 235, 235))
    draw.text((30, 200), "AMTC  AMTC  AMTC", fill=(235, 235, 235))

    # Header
    draw.rectangle([(0, 0), (width, 65)], fill="black")
    draw.text((15, 12), "AYUSH MATHEMATICS TEACHING CENTER", fill="white")
    draw.text((15, 38), "STUDENT IDENTIFICATION CARD", fill="white")

    # Design Lines
    draw.line([(0, 68), (width, 68)], fill="black", width=2)
    draw.rectangle([(15, 85), (20, 240)], fill="black")
    draw.line([(0, 255), (width, 255)], fill="black", width=1)

    # Details
    draw.text((35, 95), f"STUDENT ID :  {student_id}", fill="black")
    draw.text((35, 135), f"NAME       :  {student_name}", fill="black")
    draw.text((35, 175), f"CLASS      :  {student_class}", fill="black")

    # QR Code
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_M,
        box_size=6,
        border=1,
    )
    qr.add_data(student_id)
    qr.make(fit=True)
    qr_img = qr.make_image(fill_color="black", back_color="white").resize(
        (120, 120)
    )

    draw.rectangle([(350, 85), (478, 213)], outline="black", width=2)
    card.paste(qr_img, (354, 89))
    draw.text((368, 220), "SCAN FOR TEST", fill="black")
    draw.rectangle([(0, 0), (width - 1, height - 1)], outline="black", width=3)

    if not os.path.exists("temp_cards"):
        os.makedirs("temp_cards")

    img_path = f"temp_cards/{student_id}.png"
    card.save(img_path)

    # Print Username / ID on Terminal explicitly
    print(f"👤 Username/ID: {student_id} | Name: {student_name}")
    return img_path


# 2. Function to Combine Images into A4 Sheet PDF
def generate_a4_pdf(student_list, pdf_filename="AMTC_All_ID_Cards.pdf"):
    c = canvas.Canvas(pdf_filename, pagesize=A4)
    a4_width, a4_height = A4  # 595.27 x 841.89 points

    # Card dimensions on A4 (in points)
    card_w = 260
    card_h = 156

    # Layout Grid: 2 Columns x 3 Rows = 6 Cards per A4 Page
    margin_x = 25
    margin_y = 35
    gap_x = 25
    gap_y = 80

    cards_per_page = 6
    count = 0

    print("\n--- GENERATING ID CARDS & USERNAMES ---")
    for name, s_class in student_list:
        img_path = generate_single_card_image(name, s_class)

        # Calculate Row & Col on A4
        idx_on_page = count % cards_per_page
        col = idx_on_page % 2
        row = idx_on_page // 2

        x = margin_x + col * (card_w + gap_x)
        y = a4_height - margin_y - (row + 1) * card_h - row * gap_y

        # Draw Card Image on PDF
        c.drawImage(img_path, x, y, width=card_w, height=card_h)

        count += 1

        # New Page if current page is full (6 cards done)
        if count % cards_per_page == 0 and count < len(student_list):
            c.showPage()

    c.save()
    print("\n=======================================================")
    print(f"✅ SUCCESS: PDF Generated -> '{pdf_filename}'")
    print(
        f"📄 Total Students: {len(student_list)} | Pages Created: {(len(student_list) + 5) // 6}"
    )
    print("=======================================================")


# ==========================================
# आपके सभी 15 बच्चों की लिस्ट:
# ==========================================
students = [
    ("Ayush Kumar", "Class 4"),
    ("Monika Kumari", "Class 4"),
    ("Nishu kumari", "Class 4"),
    ("Rohan kumar", "Class 2"),
    ("Sebi kumari", "Class 2"),
    ("Deepak kumar", "Class 2"),
    ("Anjli kumari", "Class 2"),
    ("Mushkan kumari", "Class 2"),
    ("Shonakshi kumari", "Class 2"),
    ("Ladu kumar", "Class 2"),
    ("shroj kumar", "Class 2"),
    ("Ravi ranjan kumar", "Class 2"),
    ("Aaditya kumar", "Class 2"),
    ("Rishu kumar", "Class 2"),
    ("Monika Kumari", "Class 2"),
]

# Run Generator
generate_a4_pdf(students)
```
