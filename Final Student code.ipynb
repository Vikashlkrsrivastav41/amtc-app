```python
from datetime import datetime
import os
import cv2
from IPython.display import clear_output, display
import ipywidgets as widgets
import pandas as pd

# --- CONFIG & DATABASE ---
TEACHER_QR = "TEACHER-AMTC"
FILE_NAME = "Sunday_Test_Records.csv"

STUDENT_DATABASE = {
    "AMTC-101": "Ayush Kumar",
    "AMTC-102": "Monika Kumari",
    "AMTC-103": "Nishu kumari",
    "AMTC-104": "Rohan kumar",
    "AMTC-105": "Sebi kumari",
    "AMTC-106": "Deepak kumar",
    "AMTC-107": "Anjli kumari",
    "AMTC-108": "Mushkan kumari",
    "AMTC-109": "Shonakshi kumari",
    "AMTC-110": "Ladu kumar",
    "AMTC-111": "shroj kumar",
    "AMTC-112": "Ravi ranjan kumar",
    "AMTC-113": "Aaditya kumar",
    "AMTC-114": "Rishu kumar",
    "AMTC-115": "Monika Kumari",
}


# --- CAMERA SCANNER LOGIC ---
def start_continuous_scan(week_name, default_fee):
    cap = cv2.VideoCapture(0)
    detector = cv2.QRCodeDetector()
    today_date = datetime.now().strftime("%d-%m-%Y")

    # Load Existing Records to Prevent Duplicate Entry
    scanned_ids = set()
    if os.path.exists(FILE_NAME):
        df_ex = pd.read_csv(FILE_NAME)
        # Week aur Date match karke existing IDs extract karein
        existing_today = df_ex[
            (df_ex["Test_Week"] == week_name)
            & (df_ex["Attendance"] == "Present")
        ]["Student_ID"].tolist()
        scanned_ids.update(existing_today)

    print(f"📸 CAMERA LIVE ({week_name}): QR Scan चालू है...")
    print("🔴 रोकने के लिए कीबोर्ड पर 'q' दबाएं!\n")

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        data, bbox, _ = detector.detectAndDecode(frame)

        if data:
            sid = data.strip().upper()
            if sid in STUDENT_DATABASE:
                # Check Duplicate Scan
                if sid in scanned_ids:
                    print(
                        f"⚠️ ALERT: {STUDENT_DATABASE[sid]} ({sid}) की अटेंडेंस आज पहले ही लग चुकी है!"
                    )
                else:
                    scanned_ids.add(sid)
                    record = {
                        "Date": [today_date],
                        "Student_ID": [sid],
                        "Name": [STUDENT_DATABASE[sid]],
                        "Test_Week": [week_name],
                        "Attendance": ["Present"],
                        "Marks_Obtained": ["Pending"],
                        "Total_Marks": ["--"],
                        "Test_Fee": [default_fee],
                    }
                    df_new = pd.DataFrame(record)
                    if os.path.exists(FILE_NAME):
                        df_ex = pd.read_csv(FILE_NAME)
                        pd.concat([df_ex, df_new], ignore_index=True).to_csv(
                            FILE_NAME, index=False
                        )
                    else:
                        df_new.to_csv(FILE_NAME, index=False)

                    print(
                        f"✅ ATTENDANCE MARKED: {STUDENT_DATABASE[sid]} ({sid})"
                    )

        cv2.imshow("AMTC QR Scanner - Press 'q' to stop", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()
    print("\n🛑 Scanning Finished.")


def scan_single_qr():
    cap = cv2.VideoCapture(0)
    detector = cv2.QRCodeDetector()
    val = None
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        data, _, _ = detector.detectAndDecode(frame)
        if data:
            val = data.strip().upper()
            break
        cv2.imshow("Security Check - Press 'q' to cancel", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
    cap.release()
    cv2.destroyAllWindows()
    return val


# --- UI WIDGETS ---
output = widgets.Output()

# 1. Attendance Section
week_input = widgets.Text(value="Week 1", description="Test Week:")
fee_status_input = widgets.Dropdown(
    options=["Unpaid", "Paid"], value="Unpaid", description="Default Fee:"
)
btn_start_scan = widgets.Button(
    description="📹 Start Scan",
    button_style="info",
    layout=widgets.Layout(width="180px"),
)
btn_mark_absent = widgets.Button(
    description="🚫 Auto-Mark Absentees",
    button_style="warning",
    layout=widgets.Layout(width="220px"),
)

# 2. Student Info & Fee Section
student_dropdown = widgets.Dropdown(
    options=[(f"{v} ({k})", k) for k, v in STUDENT_DATABASE.items()],
    description="Student:",
)
btn_check_student = widgets.Button(
    description="🔍 View Student History",
    button_style="primary",
    layout=widgets.Layout(width="200px"),
)
fee_update_dropdown = widgets.Dropdown(
    options=["Paid", "Unpaid"], description="Fee Status:"
)
btn_update_fee = widgets.Button(
    description="💳 Update Fee Status",
    button_style="success",
    layout=widgets.Layout(width="180px"),
)

# 3. Marks Section
marks_input = widgets.IntText(value=0, description="Obtained:")
total_marks_input = widgets.IntText(value=30, description="Total Marks:")
btn_save_marks = widgets.Button(
    description="🔒 Save Marks (Teacher QR)",
    button_style="success",
    layout=widgets.Layout(width="220px"),
)

# Database Controls
btn_view_all = widgets.Button(
    description="📊 View All Records", button_style="primary"
)
btn_clear_all = widgets.Button(
    description="🗑️ Clear Database", button_style="danger"
)


# --- BUTTON LOGIC ---
def on_scan_click(b):
    with output:
        clear_output()
        start_continuous_scan(week_input.value, fee_status_input.value)


def on_mark_absent_click(b):
    with output:
        clear_output()
        today_date = datetime.now().strftime("%d-%m-%Y")
        week_name = week_input.value

        if not os.path.exists(FILE_NAME):
            print("❌ पहले अटेंडेंस स्कैन करें!")
            return

        df = pd.read_csv(FILE_NAME)
        presents = df[df["Test_Week"] == week_name]["Student_ID"].tolist()

        absent_records = []
        for sid, name in STUDENT_DATABASE.items():
            if sid not in presents:
                absent_records.append(
                    {
                        "Date": today_date,
                        "Student_ID": sid,
                        "Name": name,
                        "Test_Week": week_name,
                        "Attendance": "Absent",
                        "Marks_Obtained": "0",
                        "Total_Marks": "--",
                        "Test_Fee": "Unpaid",
                    }
                )

        if absent_records:
            df_absent = pd.DataFrame(absent_records)
            pd.concat([df, df_absent], ignore_index=True).to_csv(
                FILE_NAME, index=False
            )
            print(
                f"✅ SUCCESS: {len(absent_records)} बच्चों की 'Absent' दर्ज कर दी गई!"
            )
        else:
            print("✨ सभी बच्चे उपस्थित हैं!")


def on_check_student_click(b):
    with output:
        clear_output()
        sid = student_dropdown.value
        if os.path.exists(FILE_NAME):
            df = pd.read_csv(FILE_NAME)
            student_records = df[df["Student_ID"] == sid]
            print(f"=== 👤 RECORD FOR: {STUDENT_DATABASE[sid]} ({sid}) ===")
            if not student_records.empty:
                display(student_records)
            else:
                print("इस छात्र का कोई रिकॉर्ड नहीं मिला।")
        else:
            print("डेटाबेस खाली है!")


def on_update_fee_click(b):
    with output:
        clear_output()
        sid = student_dropdown.value
        new_status = fee_update_dropdown.value
        if os.path.exists(FILE_NAME):
            df = pd.read_csv(FILE_NAME)
            mask = df["Student_ID"] == sid
            if mask.any():
                df.loc[mask, "Test_Fee"] = new_status
                df.to_csv(FILE_NAME, index=False)
                print(
                    f"💳 {STUDENT_DATABASE[sid]} की फीस स्टेटस '{new_status}' अपडेट हो गई!"
                )
            else:
                print("❌ रिकॉर्ड नहीं मिला!")


def on_save_marks_click(b):
    with output:
        clear_output()
        print("🔒 Teacher QR स्कैन करें...")
        t_qr = scan_single_qr()
        if t_qr == TEACHER_QR:
            sid = student_dropdown.value
            m_obt = marks_input.value
            m_tot = total_marks_input.value
            if os.path.exists(FILE_NAME):
                df = pd.read_csv(FILE_NAME)
                mask = df["Student_ID"] == sid
                if mask.any():
                    df.loc[mask, "Marks_Obtained"] = str(m_obt)
                    df.loc[mask, "Total_Marks"] = str(m_tot)
                    df.to_csv(FILE_NAME, index=False)
                    print(
                        f"🎉 SAVED: {STUDENT_DATABASE[sid]} = {m_obt} / {m_tot}"
                    )
                else:
                    print("❌ अटेंडेंस रिकॉर्ड नहीं मिला!")
        else:
            print("❌ Security Check Failed!")


def on_view_all_click(b):
    with output:
        clear_output()
        if os.path.exists(FILE_NAME):
            display(pd.read_csv(FILE_NAME))


def on_clear_all_click(b):
    with output:
        clear_output()
        if os.path.exists(FILE_NAME):
            os.remove(FILE_NAME)
            print("🗑️ पूरा डेटा क्लियर हो गया!")


# Event Bindings
btn_start_scan.on_click(on_scan_click)
btn_mark_absent.on_click(on_mark_absent_click)
btn_check_student.on_click(on_check_student_click)
btn_update_fee.on_click(on_update_fee_click)
btn_save_marks.on_click(on_save_marks_click)
btn_view_all.on_click(on_view_all_click)
btn_clear_all.on_click(on_clear_all_click)

# --- DASHBOARD DISPLAY ---
display(
    widgets.VBox(
        [
            widgets.HTML(
                "<h2 style='color:#0D47A1;'>📐 AMTC Teaching Center Management System</h2>"
            ),
            widgets.HTML("<h4>1️⃣ स्कैनिंग एवं अटेंडेंस (Attendance & Scan)</h4>"),
            widgets.HBox(
                [week_input, fee_status_input, btn_start_scan, btn_mark_absent]
            ),
            widgets.HTML("<hr>"),
            widgets.HTML(
                "<h4>2️⃣ स्टूडेंट प्रोफाइल व फीस मैनेजमेंट (Student Profile & Fees)</h4>"
            ),
            widgets.HBox([student_dropdown, btn_check_student]),
            widgets.HBox([fee_update_dropdown, btn_update_fee]),
            widgets.HTML("<hr>"),
            widgets.HTML("<h4>3️⃣ मार्क्स दर्ज करें (Marks Entry)</h4>"),
            widgets.HBox([marks_input, total_marks_input, btn_save_marks]),
            widgets.HTML("<hr>"),
            widgets.HBox([btn_view_all, btn_clear_all]),
            widgets.HTML("<hr>"),
            output,
        ]
    )
)
```

```python
from datetime import datetime
import os
import time
import cv2
from IPython.display import clear_output, display
import ipywidgets as widgets
import pandas as pd

# Windows Beep Sound Support
try:
    import winsound

    def play_beep():
        winsound.Beep(2000, 400)  # Frequency 2000Hz, Duration 400ms (High Beep)

except ImportationError:

    def play_beep():
        print("\a")  # Default terminal bell for non-Windows


# --- CONFIG & DATABASE ---
TEACHER_QR = "TEACHER-AMTC"
FILE_NAME = "Sunday_Test_Records.csv"

STUDENT_DATABASE = {
    "AMTC-101": "Ayush Kumar",
    "AMTC-102": "Monika Kumari",
    "AMTC-103": "Nishu kumari",
    "AMTC-104": "Rohan kumar",
    "AMTC-105": "Sebi kumari",
    "AMTC-106": "Deepak kumar",
    "AMTC-107": "Anjli kumari",
    "AMTC-108": "Mushkan kumari",
    "AMTC-109": "Shonakshi kumari",
    "AMTC-110": "Ladu kumar",
    "AMTC-111": "shroj kumar",
    "AMTC-112": "Ravi ranjan kumar",
    "AMTC-113": "Aaditya kumar",
    "AMTC-114": "Rishu kumar",
    "AMTC-115": "Monika Kumari",
}


# --- CAMERA SCANNER LOGIC WITH SOUND & DISPLAY MESSAGE ---
def start_continuous_scan(week_name, default_fee):
    cap = cv2.VideoCapture(0)
    detector = cv2.QRCodeDetector()
    today_date = datetime.now().strftime("%d-%m-%Y")

    scanned_ids = set()
    if os.path.exists(FILE_NAME):
        df_ex = pd.read_csv(FILE_NAME)
        existing_today = df_ex[
            (df_ex["Test_Week"] == week_name)
            & (df_ex["Attendance"] == "Present")
        ]["Student_ID"].tolist()
        scanned_ids.update(existing_today)

    print(f"📸 CAMERA LIVE ({week_name}): Scan Chalu Hai...")
    print("🔴 Rokne Ke Liye 'q' Dabayein!\n")

    msg_display = ""
    msg_color = (0, 255, 0)
    msg_timer = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # Display Message on Camera Feed
        if time.time() < msg_timer:
            cv2.putText(
                frame,
                msg_display,
                (20, 60),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                msg_color,
                2,
            )

        data, bbox, _ = detector.detectAndDecode(frame)

        if data:
            sid = data.strip().upper()
            if sid in STUDENT_DATABASE:
                if sid in scanned_ids:
                    # Pehle Se Lag Chuki Hai Warning
                    msg_display = f"ALREADY MARKED: {STUDENT_DATABASE[sid]}"
                    msg_color = (0, 0, 255)  # Red Color
                    msg_timer = time.time() + 2  # Show for 2 seconds
                else:
                    # New Attendance Done!
                    scanned_ids.add(sid)
                    student_name = STUDENT_DATABASE[sid]

                    # 1. Play Beep Sound (🔊 आवाज़ करेगा)
                    play_beep()

                    # 2. Save in CSV
                    record = {
                        "Date": [today_date],
                        "Student_ID": [sid],
                        "Name": [student_name],
                        "Test_Week": [week_name],
                        "Attendance": ["Present"],
                        "Marks_Obtained": ["Pending"],
                        "Total_Marks": ["--"],
                        "Test_Fee": [default_fee],
                    }
                    df_new = pd.DataFrame(record)
                    if os.path.exists(FILE_NAME):
                        df_ex = pd.read_csv(FILE_NAME)
                        pd.concat([df_ex, df_new], ignore_index=True).to_csv(
                            FILE_NAME, index=False
                        )
                    else:
                        df_new.to_csv(FILE_NAME, index=False)

                    # 3. Print in Terminal & Show Green Message on Camera Screen
                    print(
                        f"🔊 [BEEP!] ATTENDANCE SUCCESS: {student_name} ({sid})"
                    )
                    msg_display = f"SUCCESS: {student_name} Ka Attendance Ho Gaya!"
                    msg_color = (0, 255, 0)  # Green Color
                    msg_timer = time.time() + 3  # Show for 3 seconds

        cv2.imshow("AMTC QR Scanner - Press 'q' to Stop", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()
    print("\n🛑 Scanning Finished.")


def scan_single_qr():
    cap = cv2.VideoCapture(0)
    detector = cv2.QRCodeDetector()
    val = None
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        data, _, _ = detector.detectAndDecode(frame)
        if data:
            val = data.strip().upper()
            play_beep()  # Beep sound on teacher QR too
            break
        cv2.imshow("Teacher Verification - Press 'q' to cancel", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
    cap.release()
    cv2.destroyAllWindows()
    return val


# --- UI WIDGETS ---
output = widgets.Output()

week_input = widgets.Text(value="Week 1", description="Test Week:")
fee_status_input = widgets.Dropdown(
    options=["Unpaid", "Paid"], value="Unpaid", description="Default Fee:"
)
btn_start_scan = widgets.Button(
    description="📹 Start Scan (With Sound)",
    button_style="info",
    layout=widgets.Layout(width="220px"),
)
btn_mark_absent = widgets.Button(
    description="🚫 Auto-Mark Absentees",
    button_style="warning",
    layout=widgets.Layout(width="200px"),
)

student_dropdown = widgets.Dropdown(
    options=[(f"{v} ({k})", k) for k, v in STUDENT_DATABASE.items()],
    description="Student:",
)
btn_check_student = widgets.Button(
    description="🔍 View Student History",
    button_style="primary",
    layout=widgets.Layout(width="200px"),
)
fee_update_dropdown = widgets.Dropdown(
    options=["Paid", "Unpaid"], description="Fee Status:"
)
btn_update_fee = widgets.Button(
    description="💳 Update Fee Status",
    button_style="success",
    layout=widgets.Layout(width="180px"),
)

marks_input = widgets.IntText(value=0, description="Obtained:")
total_marks_input = widgets.IntText(value=30, description="Total Marks:")
btn_save_marks = widgets.Button(
    description="🔒 Save Marks (Teacher QR)",
    button_style="success",
    layout=widgets.Layout(width="220px"),
)

btn_view_all = widgets.Button(
    description="📊 View All Records", button_style="primary"
)
btn_clear_all = widgets.Button(
    description="🗑️ Clear Database", button_style="danger"
)


# --- BUTTON LOGIC ---
def on_scan_click(b):
    with output:
        clear_output()
        start_continuous_scan(week_input.value, fee_status_input.value)


def on_mark_absent_click(b):
    with output:
        clear_output()
        today_date = datetime.now().strftime("%d-%m-%Y")
        week_name = week_input.value

        if not os.path.exists(FILE_NAME):
            print("❌ पहले अटेंडेंस स्कैन करें!")
            return

        df = pd.read_csv(FILE_NAME)
        presents = df[df["Test_Week"] == week_name]["Student_ID"].tolist()

        absent_records = []
        for sid, name in STUDENT_DATABASE.items():
            if sid not in presents:
                absent_records.append(
                    {
                        "Date": today_date,
                        "Student_ID": sid,
                        "Name": name,
                        "Test_Week": week_name,
                        "Attendance": "Absent",
                        "Marks_Obtained": "0",
                        "Total_Marks": "--",
                        "Test_Fee": "Unpaid",
                    }
                )

        if absent_records:
            df_absent = pd.DataFrame(absent_records)
            pd.concat([df, df_absent], ignore_index=True).to_csv(
                FILE_NAME, index=False
            )
            print(
                f"✅ SUCCESS: {len(absent_records)} बच्चों की 'Absent' दर्ज कर दी गई!"
            )
        else:
            print("✨ सभी बच्चे उपस्थित हैं!")


def on_check_student_click(b):
    with output:
        clear_output()
        sid = student_dropdown.value
        if os.path.exists(FILE_NAME):
            df = pd.read_csv(FILE_NAME)
            student_records = df[df["Student_ID"] == sid]
            print(f"=== 👤 RECORD FOR: {STUDENT_DATABASE[sid]} ({sid}) ===")
            if not student_records.empty:
                display(student_records)
            else:
                print("इस छात्र का कोई रिकॉर्ड नहीं मिला।")
        else:
            print("डेटाबेस खाली है!")


def on_update_fee_click(b):
    with output:
        clear_output()
        sid = student_dropdown.value
        new_status = fee_update_dropdown.value
        if os.path.exists(FILE_NAME):
            df = pd.read_csv(FILE_NAME)
            mask = df["Student_ID"] == sid
            if mask.any():
                df.loc[mask, "Test_Fee"] = new_status
                df.to_csv(FILE_NAME, index=False)
                print(
                    f"💳 {STUDENT_DATABASE[sid]} की फीस स्टेटस '{new_status}' अपडेट हो गई!"
                )
            else:
                print("❌ रिकॉर्ड नहीं मिला!")


def on_save_marks_click(b):
    with output:
        clear_output()
        print("🔒 Teacher QR स्कैन करें...")
        t_qr = scan_single_qr()
        if t_qr == TEACHER_QR:
            sid = student_dropdown.value
            m_obt = marks_input.value
            m_tot = total_marks_input.value
            if os.path.exists(FILE_NAME):
                df = pd.read_csv(FILE_NAME)
                mask = df["Student_ID"] == sid
                if mask.any():
                    df.loc[mask, "Marks_Obtained"] = str(m_obt)
                    df.loc[mask, "Total_Marks"] = str(m_tot)
                    df.to_csv(FILE_NAME, index=False)
                    print(
                        f"🎉 SAVED: {STUDENT_DATABASE[sid]} = {m_obt} / {m_tot}"
                    )
                else:
                    print("❌ अटेंडेंस रिकॉर्ड नहीं मिला!")
        else:
            print("❌ Security Check Failed!")


def on_view_all_click(b):
    with output:
        clear_output()
        if os.path.exists(FILE_NAME):
            display(pd.read_csv(FILE_NAME))


def on_clear_all_click(b):
    with output:
        clear_output()
        if os.path.exists(FILE_NAME):
            os.remove(FILE_NAME)
            print("🗑️ पूरा डेटा क्लियर हो गया!")


# Event Bindings
btn_start_scan.on_click(on_scan_click)
btn_mark_absent.on_click(on_mark_absent_click)
btn_check_student.on_click(on_check_student_click)
btn_update_fee.on_click(on_update_fee_click)
btn_save_marks.on_click(on_save_marks_click)
btn_view_all.on_click(on_view_all_click)
btn_clear_all.on_click(on_clear_all_click)

# --- DASHBOARD DISPLAY ---
display(
    widgets.VBox(
        [
            widgets.HTML(
                "<h2 style='color:#0D47A1;'>📐 AMTC Teaching Center System (With Sound)</h2>"
            ),
            widgets.HTML("<h4>1️⃣ स्कैनिंग एवं अटेंडेंस (Attendance & Scan)</h4>"),
            widgets.HBox(
                [week_input, fee_status_input, btn_start_scan, btn_mark_absent]
            ),
            widgets.HTML("<hr>"),
            widgets.HTML(
                "<h4>2️⃣ स्टूडेंट प्रोफाइल व फीस मैनेजमेंट (Student Profile & Fees)</h4>"
            ),
            widgets.HBox([student_dropdown, btn_check_student]),
            widgets.HBox([fee_update_dropdown, btn_update_fee]),
            widgets.HTML("<hr>"),
            widgets.HTML("<h4>3️⃣ मार्क्स दर्ज करें (Marks Entry)</h4>"),
            widgets.HBox([marks_input, total_marks_input, btn_save_marks]),
            widgets.HTML("<hr>"),
            widgets.HBox([btn_view_all, btn_clear_all]),
            widgets.HTML("<hr>"),
            output,
        ]
    )
)
```
